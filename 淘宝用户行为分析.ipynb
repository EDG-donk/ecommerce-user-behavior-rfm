每天活跃度变化
  计算出日访问量，日独立访客量和人均访问量，封装成一个新df

```python
#日访问量 pv page views
pv_daily_s=data.groupby(by='date')['user_id'].count()
pv_daily_s.head()
```

```python
#日独立访客量 uv unique visitors
uv_daily_s=data.groupby(by='date')['user_id'].nunique()
uv_daily_s.head()
```

```python
#人均访问量(日访问量/日独立访客量)
uv_pv_s=pv_daily_s/uv_daily_s
uv_pv_s.head()
```

```python
#形成新的数据集
df=pd.concat((pv_daily_s,uv_daily_s,uv_pv_s),axis=1)
df.columns=['pv','uv','pv/uv']
df.head()
```

```python
#绘制pv折线图
plt.plot(df.index,df.pv)
plt.title('pv')
plt.xticks(rotation=30)
plt.show()
```

```python
#绘制uv折线图
plt.plot(df.index,df.uv)
plt.title('uv')
plt.xticks(rotation=30)
plt.show()
```

UV在一阶段升高,但是PV降低了,说明网站的弹出率升高了,需要对网站那些人进行观察分析,是什么原因导致弹出
  1.用户没有看到想看的东西就走了。说明关键词和着陆页相关性较低
  2.着陆页体验太差,用户可能在网站看到想要的东西,但是浏览选择不顺畅
  3.可能是网页打卡速度太慢

```python
#绘制pv/uv折线图
plt.plot(df.index,df['pv/uv'])
plt.title('pv/uv')
plt.xticks(rotation=30)
plt.show()
```

结论:在12月5日之前,活跃度在一定水平波动。12月5日后,活跃度开始明显上升,并在双十二当天达到峰值
可能原因:12月5日之后双十二预热活动开始,用户活跃度上升

用户价值分析(RMF模型)
  对已购用户进行价值划分
  各类用户占比

```python
#将已购用户的数据单独取出
buy_df=data.loc[data['behavior_type']==4]
buy_df.head()
```

```python
buy_df['date'].max()
```

```python
list(buy_df.groupby(by='user_id')['date'])[3]
```

```python
#计算R:R表示客户最近一次交易时间的间隔
now_date=buy_df['date'].max()
R=buy_df.groupby(by='user_id')['date'].apply(lambda x:now_date - x.max())/ np.timedelta64(1,'D')
```

```python
R.head()
```

```python
#计算F:每个用户消费频率
F= buy_df.groupby(by='user_id')['date'].count()
```

```python
F.head()
```

```python
pd.DataFrame(data=[R,F],index=['R','F'])
```

```python
rfm=pd.DataFrame(data=[R,F],index=['R','F']).T
rfm.head()
```

```python
#将各维度分成两个程度
recent_avg=rfm['R'].mean()
freq_avg=rfm['F'].mean()
# R应该是越小越好，则R小于均值返回1，否则返回0
def rec_value(x):
    if x < recent_avg:  # R avg
        return '1'
    else:
        return '0'
# F应该是越大越好，则F大于均值返回1，否则返回0
def freq_value(x):
    if x > freq_avg:
        return '1'
    else:
        return '0'
rfm['R_value'] = rfm['R'].apply(rec_value)
rfm['F_value'] = rfm['F'].apply(freq_value)
```

```python
#将R和F拼接到一起
rfm['rfm']=rfm['R_value'].str.cat(rfm['F_value'])
```

```python
rfm['rfm'].head()
```

```python
#根据R和F的拼接判定用户等级
def rfm_value(x):
    if x == '10':   #购买间隔短，但是购买频率低
        return '重要发展客户'
    elif x == '01': #购买间隔长，但是购买频率高
        return '重要保持客户'
    elif x == '00':
        return '重要挽留客户'
    else:
        return '重要价值客户'
rfm['user_type'] = rfm['rfm'].apply(rfm_value)
rfm.head()
```

```python
#各用户占比
user_type_count_s=rfm['user_type'].value_counts()
user_type_count_s/user_type_count_s.sum()*100
```

结论
  重要挽留客户:占比最大,该类用户消费时间间隔较远,并且消费频次低,需要主动联系客户,调查问题在哪,可以通过短信,邮件联系,尽可能减少流失
  重要发展客户:消费频次低,可以适当给点折扣增加用户购买频率
  重要价值客户:重点用户,用户数量较少,可以适当提供一些VIP服务
  重要保持客户:占比最小,该类用户消费时间较远,但是消费频次高,该类用户可能一次性购买很多东西,对于这类用户,也可以主动联系,关注购买习惯,做好精准营销,满足该类用户需求
